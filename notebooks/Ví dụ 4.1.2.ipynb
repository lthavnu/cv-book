```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Đọc ảnh mức xám
img = cv2.imread("cameraman.png", cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError("Không tìm thấy file cameraman.png")

# Các giá trị gamma
gamma_values = [0, 1, 2, 3]

# Hiển thị ảnh gốc + các ảnh sau lọc
plt.figure(figsize=(15, 6))

# Ảnh gốc
plt.subplot(1, len(gamma_values) + 1, 1)
plt.imshow(img, cmap='gray')
plt.title("Ảnh gốc")
plt.axis("off")

# Lọc Gaussian với kernel 5x4
for i, gamma in enumerate(gamma_values, start=2):
    filtered = cv2.GaussianBlur(
        img,
        ksize=(5, 5),   # width x height = 4 x 5 tương ứng cửa sổ 5x4
        sigmaX=gamma,
        sigmaY=gamma
    )

    plt.subplot(1, len(gamma_values) + 1, i)
    plt.imshow(filtered, cmap='gray')
    plt.title(f"gamma = {gamma}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
